**Data Cleaning and Outlier Handling**

Data cleaning is an important step in preparing the dataset for accurate analysis. In this project, Python and Pandas were used to clean the airline dataset and improve data quality.

The following data cleaning steps were performed:

1. **Handling Missing Values:** Missing values in numerical columns, including Flight_Distance, Departure_Delay, Arrival_Delay, and Flight_Duration, were filled using the median. Missing values in the Delay_Reason column were filled with the category 'Late Aircraft'.

2. **Removing Duplicate Records:** Duplicate records were identified and removed to avoid repeated data. After removing 10 duplicate records, the dataset contained 1,990 rows and 10 columns.

3. **Outlier Detection:** The Interquartile Range (IQR) method was used to detect outliers in numerical columns. Initially, 9 outliers were identified in Departure_Delay and 14 in Arrival_Delay.

4. **Outlier Handling:** The identified outliers in Departure_Delay and Arrival_Delay were capped using the lower and upper limits calculated by the IQR method. This helped reduce the influence of extreme delay values on the analysis.

5. **Final Data Validation:** After cleaning, the dataset was checked to ensure that no missing values or duplicate records remained. The cleaned dataset contained 1,990 rows and 10 columns.

These data cleaning steps helped prepare the dataset for exploratory data analysis and visualization. However, filling missing delay reasons with 'Late Aircraft' may affect the distribution of delay categories and is a limitation of this analysis.

#Outlier Analysis


```python
numeric_columns = ['Flight_Distance','Departure_Delay','Arrival_Delay','Flight_Duration']
a[numeric_columns].describe()
```

```python
for col in numeric_columns:
  Q1 = a[col].quantile(0.25)
  Q3= a[col].quantile(0.75)
  IQR = Q3-Q1

  lower = Q1 - 1.5* IQR
  upper =  Q3 + 1.5 * IQR

  outliers = a[(a[col]<lower) | (a[col]>upper)]
  print(col,":",len(outliers),"outliers")
```

#Outlier Handling


```python
for col in ['Departure_Delay', 'Arrival_Delay']:
    Q1 = a[col].quantile(0.25)
    Q3 = a[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    a[col] = a[col].clip(lower=lower, upper=upper)
```

```python
for coll in ['Departure_Delay','Arrival_Delay']:
  Q1 = a[col].quantile(0.25)
  Q3 = a[col].quantile(0.75)
  IQR = Q3-Q1

  lower =Q1 - 1.5 *IQR
  upper =Q3 + 1.5*IQR

  outliers = a[(a[col] < lower)| (a[col] > upper)]
  print(col,":",len(outliers),"outliers")
```
